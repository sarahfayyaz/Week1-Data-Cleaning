**Imported Pandas:**

 Used Pandas to load, clean, and analyze the dataset.


**Loaded the dataset:**

 Read the CSV file and checked its shape, first rows, data types, missing values, duplicates, and summary statistics to understand data quality.

```python
import  pandas as pd
file_path = "/content/sample_data/messy_applicant_dataset.csv"

df = pd.read_csv(file_path)
df.shape
df.head()
df.info()
df.isnull().sum()
df.duplicated().sum()
df.describe()
df.describe(include='object')
```

**Handled missing values:**

Dropped rows with missing Applicant Name because applicant identity is important. Filled missing University with Not Specified and Status with Under Review to preserve useful records.

```python
df = df.dropna(subset=['Applicant Name'])
df['University'] = df['University'].fillna('Not Specified')
df['Status'] = df['Status'].fillna('Under Review')
```

**Removed duplicates:**

 Removed exact duplicate rows and duplicate emails, keeping the first record to avoid repeated applicant entries.

```python
# Check exact duplicates first
df.duplicated().sum()
# Remove exact duplicate rows
df = df.drop_duplicates()
df = df.drop_duplicates(subset=['Email'], keep='first')
```

**Standardized Domain Applied: **

Mapped variations such as web dev, webdev, and web development to Web Development for consistency.

```python
domain_mapping = {      'web dev': 'Web Development',
                  'webdev': 'Web Development',
                        'web development': 'Web Development',  }
df['Domain Applied'] = df['Domain Applied'].map(domain_mapping)
df['Domain Applied'].unique()
```

**Cleaned Status:**

Used .str.strip().str.title() to remove extra spaces and

*   List item
*   List item

standardize capitalization.

```python
df['Status'] = df['Status'].str.strip().str.title()
print("\nMissing values after cleaning:")
print(df.isnull().sum())
df.to_csv('/content/sample_data/messy_applicant_dataset.csv',
          index=False)

print("\nCleaned dataset saved successfully!")
print(df.head())
```

**Converted Application Date:**

Used pd.to_datetime(errors='coerce') to convert dates into a proper datetime format; invalid dates become NaT.

**Cleaned Phone numbers:**

Removed hyphens and spaces to make phone numbers more consistent.

```python
df['Application Date'] = pd.to_datetime(df['Application Date'],
errors='coerce')
df['Phone'] = df['Phone'].str.replace('-', '', regex=False).str.replace(' ',
'', regex=False)
df.info()
```

**Verified and saved: **

Checked missing values and data types again, then saved the cleaned dataset as a CSV file.

```python
df.to_csv('applicants_cleaned.csv', index=False)
```
